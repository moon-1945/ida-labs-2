## Extract webpage data given the url

Create a Python script that performs basic web scraping on a page to extract all the information into text and returns it as a string. String should not contain tags.

```python
import re
import sys
 
import requests
from bs4 import BeautifulSoup
 
HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"}
 
# Elements whose content is not visible text
NON_TEXT_TAGS = ["script", "style", "noscript", "template", "head", "svg", "iframe"]
 
 
def parse_web_page(url, timeout=10):
    """
    Fetch a web page and return its text content without any HTML tags.
 
    Args:
        url (str): The URL of the web page to fetch.
        timeout (int): Seconds to wait for the server before giving up.
 
    Returns:
        str: The visible text of the page, one block of text per line.
 
    Raises:
        requests.HTTPError: If the HTTP request returned an unsuccessful status code.
    """
    response = requests.get(url, headers=HEADERS, timeout=timeout)
    response.raise_for_status()
 
    # Let requests guess the encoding when the server doesn't declare one
    if not response.encoding or response.encoding.lower() == "iso-8859-1":
        response.encoding = response.apparent_encoding
 
    soup = BeautifulSoup(response.text, "html.parser")
 
    # Remove elements that never contain readable text
    for tag in soup(NON_TEXT_TAGS):
        tag.decompose()
 
    # Pull out the text, one line per block element
    text = soup.get_text(separator="\n")
 
    # Normalize whitespace: trim each line, drop empty lines, squeeze repeated spaces
    lines = (re.sub(r"[ \t\xa0]+", " ", line).strip() for line in text.splitlines())
    return "\n".join(line for line in lines if line)

"""
    Replace the following links with links you wish to parse for this assignemnt.
"""
print(parse_web_page('https://en.wikipedia.org/wiki/Wikipedia')[:255])
print(parse_web_page('https://github.com/emmastrienko'))
```

## Parse the json file

Parse the `weather.json` file and return weather data for a specific date, that is given as a parameter. Return the data as an array.

```python
import json
from datetime import datetime


def parse_json(date):
    """
    Parse the data from weather.json file and return weather data for a given date.

    Args:
        date (str): The date for which we look up the weather.

    Returns:
        list: a list of weather data for a given date.
    """
    target = datetime.strptime(date, "%Y-%m-%d").date()

    with open("resources/weather.json", encoding="utf-8") as f:
        data = json.load(f)

    return [
        day for day in data["daily"]
        if datetime.strptime(day["date"], "%Y-%m-%d").date() == target
    ]


target_date = '2024-8-19'
print(parse_json(target_date))
```

## Parse the csv file

Parse the `weather.csv` file and return weather data for a specific date, that is given as a parameter. Return the data as an array.

```python
import csv
from datetime import datetime
from pathlib import Path




def parse_csv(date):
    """
    Parse the data from weather.csv file and return weather data for a given date.

    Args:
        date (str): The date for which we look up the weather.

    Returns:
        list: a list of weather data for a given date.
    """
    target = datetime.strptime(date, "%Y-%m-%d").date()
    result = []

    with open("resources/weather.csv", encoding="utf-8", newline="") as f:
        reader = csv.DictReader(f)
        reader.fieldnames = [name.strip() for name in reader.fieldnames]

        for row in reader:
            row_date = datetime.strptime(row["CET"], "%Y-%m-%d").date()
            if row_date == target:
                result.append(row)

    return result


target_date = '1997-5-22'
print(parse_csv(target_date))
```

## Data imputation and normalization.

Download kaggle dataset with incomplete/missing data, and try data imputation techniques (mean, median, mode, knn) on that data. Normalize the data with best option for each dataset (minmax, z-score, robust). Visualize the data (choose your own approach to data visualization).

```python
"""Load dataset_mar.csv from Kaggle, impute missing values, normalize it and visualize it.

The dataset has three columns:
    Patient_ID      unique identifier (a label, not a feature)
    Age             complete
    Blood_Pressure  has gaps that depend on Age (MAR: Missing At Random)

Install dependencies:
    pip install kagglehub[pandas-datasets] pandas numpy matplotlib

Usage:
    python dataset_pipeline.py [csv_file_name] [output_dir]
"""

import logging
import sys
from pathlib import Path

import kagglehub
import matplotlib

matplotlib.use("Agg")  # save figures to files, no window needed
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from kagglehub import KaggleDatasetAdapter

DATASET_URL = "https://www.kaggle.com/datasets/suryaprabha19/different-types-of-missing-data"
CSV_FILE = "dataset_mar.csv"
ID_COLUMN = "Patient_ID"
PLOTS_DIR = "plots"

logger = logging.getLogger(__name__)


def _feature_columns(df):
    """Numeric columns that carry data (the ID column is only a label)."""
    return [c for c in df.select_dtypes(include=np.number).columns if c != ID_COLUMN]


def load_dataset(url, csv_file):
    """
    Load the dataset from Kaggle using kagglehub.

    Args:
        url (str): The URL of the Kaggle dataset.
        csv_file (str): The name of the CSV file to load from the dataset.
    Returns:
        pd.DataFrame: The loaded dataset as a pandas DataFrame.
    """
    # kagglehub expects a handle like "owner/dataset", not the full URL
    handle = url.split("/datasets/")[-1].strip("/")

    df = kagglehub.dataset_load(
        KaggleDatasetAdapter.PANDAS,
        handle,
        csv_file,
    )
    return df


def impute_data(df):
    """
    Parse the data from file and impute missing values using your preferred method.

    Method: regression imputation. The gaps in Blood_Pressure depend on Age (MAR),
    so filling them with one global mean or median would bias the result. Instead,
    each column with gaps is predicted by linear regression from the columns that
    are complete (here: Age). Predictions are clipped to the observed range.
    Non-numeric columns, if any, are filled with the most frequent value.
    Columns that are completely empty are dropped.
    Progress and decisions are reported with the logging module (INFO level).

    Args:
        df (pd.DataFrame): The dataset as a pandas DataFrame.

    Returns:
        pd.DataFrame: The dataset with imputed values.
    """
    result = df.copy()
    logger.info("Imputation started: %d rows, %d columns", *result.shape)

    empty_cols = result.columns[result.isna().all()].tolist()
    if empty_cols:
        logger.warning("Dropping completely empty columns: %s", empty_cols)
        result = result.drop(columns=empty_cols)

    features = _feature_columns(result)
    complete = [c for c in features if not result[c].isna().any()]
    logger.debug("Feature columns: %s | complete columns: %s", features, complete)

    for col in features:
        missing = result[col].isna()
        n_missing = int(missing.sum())
        if n_missing == 0:
            continue

        known = ~missing
        predictors = [c for c in complete if c != col]
        logger.info(
            "Column '%s': %d missing values (%.1f%%)", col, n_missing, 100 * n_missing / len(result)
        )

        if not predictors:  # nothing to regress on, fall back to the median
            median = result.loc[known, col].median()
            logger.warning(
                "Column '%s': no complete columns to regress on, filling with median %.3f", col, median
            )
            result.loc[missing, col] = median
            continue

        # Least-squares fit on rows where the value is known: col ~ predictors
        x_known = np.column_stack(
            [np.ones(known.sum()), result.loc[known, predictors].to_numpy(float)]
        )
        coef, *_ = np.linalg.lstsq(x_known, result.loc[known, col].to_numpy(float), rcond=None)
        logger.info(
            "Column '%s': regression fitted on %d known rows, predictors=%s, intercept=%.3f, coefficients=%s",
            col,
            int(known.sum()),
            predictors,
            coef[0],
            {p: round(float(c), 4) for p, c in zip(predictors, coef[1:])},
        )

        # Predict the missing rows and keep the values inside the observed range
        x_missing = np.column_stack(
            [np.ones(missing.sum()), result.loc[missing, predictors].to_numpy(float)]
        )
        low, high = result.loc[known, col].min(), result.loc[known, col].max()
        raw_prediction = x_missing @ coef
        predicted = np.clip(raw_prediction, low, high)

        n_clipped = int((raw_prediction != predicted).sum())
        if n_clipped:
            logger.warning(
                "Column '%s': %d predictions were outside the observed range [%.3f, %.3f] and were clipped",
                col,
                n_clipped,
                low,
                high,
            )

        result.loc[missing, col] = predicted
        logger.info(
            "Column '%s': imputed %d values, mean of imputed values %.3f (mean of known values %.3f)",
            col,
            n_missing,
            predicted.mean(),
            result.loc[known, col].mean(),
        )

    for col in result.select_dtypes(exclude=np.number).columns:
        n_missing = int(result[col].isna().sum())
        if n_missing == 0:
            continue

        mode = result[col].mode(dropna=True)
        fill_value = mode.iloc[0] if not mode.empty else "Unknown"
        logger.info(
            "Column '%s': %d missing values filled with the most frequent value %r",
            col,
            n_missing,
            fill_value,
        )
        result[col] = result[col].fillna(fill_value)

    logger.info("Imputation finished: %d missing values left", int(result.isna().sum().sum()))
    return result


def normalize_data(df):
    """
    Normalize the data using your preferred method.

    Method: min-max scaling of Age and Blood_Pressure to the range [0, 1].
    Constant columns become 0. Patient_ID is an identifier and is left unchanged.

    Args:
        df (pd.DataFrame): The dataset as a pandas DataFrame.

    Returns:
        pd.DataFrame: The dataset with normalized values.
    """
    result = df.copy()

    for col in _feature_columns(result):
        col_min, col_max = result[col].min(), result[col].max()
        span = col_max - col_min
        result[col] = 0.0 if span == 0 else (result[col] - col_min) / span

    return result


def visualize_data(df, original=None, output_dir=PLOTS_DIR, prefix=""):
    """
    Visualize the data using Matplotlib. Use more then one visualization.
    Save the results in the repository.

    Saved figures (a plot is skipped when the data does not support it):
        missing_values:   share of missing values per column
        missing_pattern:  where the missing values are, row by column
        mar_check_*:      distribution of Age for rows with and without Blood_Pressure,
                          which shows that the gaps depend on Age (MAR)
        scatter_*:        Age against Blood_Pressure; imputed points are red
        histograms:       distribution of every feature
        boxplots:         spread and outliers of every feature
        correlation:      correlation matrix of the features

    Args:
        df (pd.DataFrame): The dataset as a pandas DataFrame.
        original (pd.DataFrame, optional): The data before imputation. Pass it when
            df is imputed or normalized, so that the originally missing rows can be
            highlighted. When omitted, df itself is used to find the gaps.
        output_dir (str | Path): Folder where the images are saved.
        prefix (str): Text added to the start of every file name.

    Returns:
        None: None.
    """
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    features = _feature_columns(df)
    reference = df if original is None else original

    def save(fig, name):
        fig.tight_layout()
        fig.savefig(output_dir / f"{prefix}{name}.png", dpi=150)
        plt.close(fig)

    # Columns that had gaps, and the complete columns that can explain them
    gap_cols = [c for c in features if c in reference and reference[c].isna().any()]
    complete = [c for c in features if c in reference and not reference[c].isna().any()]

    # 1. Missing values per column (only meaningful before imputation)
    missing_share = reference.isna().mean() * 100
    if original is None and missing_share.any():
        fig, ax = plt.subplots(figsize=(7, 3.5))
        missing_share.sort_values().plot.barh(ax=ax, color="tab:red")
        ax.set_xlabel("Missing values, %")
        ax.set_title("Missing values per column")
        save(fig, "missing_values")

        # 2. Missing-value pattern
        fig, ax = plt.subplots(figsize=(7, 5))
        ax.imshow(reference.isna().to_numpy(), aspect="auto", cmap="Greys", interpolation="nearest")
        ax.set_xticks(range(len(reference.columns)))
        ax.set_xticklabels(reference.columns, rotation=45)
        ax.set_ylabel("Row")
        ax.set_title("Missing-value pattern (dark = missing)")
        save(fig, "missing_pattern")

    for y in gap_cols:
        gap = reference[y].isna()

        for x in complete:
            # 3. MAR check: do rows with a gap differ from rows without one?
            if original is None:
                bins = np.histogram_bin_edges(df[x].dropna(), bins=15)
                fig, ax = plt.subplots(figsize=(7, 4))
                ax.hist(df.loc[~gap, x], bins=bins, alpha=0.6, density=True, label=f"{y} observed")
                ax.hist(df.loc[gap, x], bins=bins, alpha=0.6, density=True, label=f"{y} missing")
                ax.set_xlabel(x)
                ax.set_ylabel("Density")
                ax.set_title(f"{x} of rows with and without {y}")
                ax.legend()
                save(fig, f"mar_check_{x}")

            # 4. Scatter, imputed points highlighted
            fig, ax = plt.subplots(figsize=(7, 5))
            ax.scatter(df.loc[~gap, x], df.loc[~gap, y], s=12, alpha=0.5, label="Observed")
            if original is not None:
                ax.scatter(df.loc[gap, x], df.loc[gap, y], s=18, color="tab:red", label="Imputed")
            ax.set_xlabel(x)
            ax.set_ylabel(y)
            ax.set_title(f"{y} against {x}")
            ax.legend()
            save(fig, f"scatter_{x}_{y}")

    if features:
        # 5. Histograms
        fig, axes = plt.subplots(1, len(features), figsize=(5 * len(features), 4), squeeze=False)
        for ax, col in zip(axes[0], features):
            ax.hist(df[col].dropna(), bins=20, color="tab:blue", edgecolor="white")
            ax.set_title(col)
            ax.set_ylabel("Count")
        fig.suptitle("Distributions")
        save(fig, "histograms")

        # 6. Boxplots (one panel per feature, since the scales differ)
        fig, axes = plt.subplots(1, len(features), figsize=(3.5 * len(features), 4), squeeze=False)
        for ax, col in zip(axes[0], features):
            ax.boxplot(df[col].dropna())
            ax.set_xticks([1])
            ax.set_xticklabels([col])
        fig.suptitle("Boxplots")
        save(fig, "boxplots")

    if len(features) >= 2:
        # 7. Correlation matrix
        corr = df[features].corr()
        fig, ax = plt.subplots(figsize=(5, 4.5))
        im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
        ax.set_xticks(range(len(corr)))
        ax.set_yticks(range(len(corr)))
        ax.set_xticklabels(corr.columns, rotation=45)
        ax.set_yticklabels(corr.columns)
        for i in range(len(corr)):
            for j in range(len(corr)):
                ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")
        fig.colorbar(im, ax=ax, label="Correlation")
        ax.set_title("Correlation matrix")
        save(fig, "correlation")


def main(csv_file=CSV_FILE, output_dir=PLOTS_DIR):
    raw = load_dataset(DATASET_URL, csv_file)
    print("Missing values per column before imputation:")
    print(raw.isna().sum(), end="\n\n")

    visualize_data(raw, output_dir=output_dir, prefix="raw_")  # shows where and why data is missing

    imputed = impute_data(raw)
    visualize_data(imputed, original=raw, output_dir=output_dir, prefix="imputed_")

    normalized = normalize_data(imputed)
    visualize_data(normalized, original=raw, output_dir=output_dir, prefix="normalized_")

    print(f"Missing values after imputation: {int(imputed.isna().sum().sum())}")
    print(f"Plots saved to: {output_dir}")



logging.basicConfig(
    level=logging.INFO,  # use logging.DEBUG for more detail
    format="%(asctime)s | %(levelname)-7s | %(message)s",
    datefmt="%H:%M:%S",
)
main()
```
